## Part 1 — Theory Recap

### How to Read a pvT Graph
A pvT (pressure–volume–Temperature) graph plots **specific volume** [cm³/g] on the Y-axis  
against **temperature** [°C] on the X-axis, with one curve per pressure.

To find shrinkage from a pvT graph:
1. Locate the **gate-freeze temperature** $T_{\text{freeze}}$ on the X-axis.
2. Read the **specific volume at processing** $v_{\text{proc}}$ at $T_{\text{freeze}}$ and melt pressure.
3. Read the **specific volume at ambient** $v_{\text{solid}}$ at ambient temperature (typically 23°C) and atmospheric pressure.

### Volumetric Shrinkage (Q11)
$$S_{\text{vol}} = \frac{v_{\text{proc}} - v_{\text{solid}}}{v_{\text{proc}}} \times 100\%$$

**Typical values:**  PP ≈ 6–8 %,  PC-ABS ≈ 2–4 %,  POM ≈ 8–10 %,  PA66 ≈ 3–5 %

### Linear Shrinkage (Q12)
Assuming isotropic shrinkage in all three directions:
$$S_{\text{lin}} \approx \frac{S_{\text{vol}}}{3}$$

**Example:** $S_{\text{vol}} = 6\%$ → $S_{\text{lin}} \approx 2\%$

### Tool Cavity Dimension
$$M_d = \frac{P_d}{1 - S_{\text{lin}}/100}$$

where $P_d$ is the desired part dimension at ambient temperature.

---

## ⚠ Common Exam Pitfalls
1. **Volumetric vs linear:** Always confirm whether the question asks for volumetric (6–8%) or linear (~2%). A very common mistake is to confuse the two.
2. **Reading the right curve:** Use the **melt pressure** curve at $T_{\text{freeze}}$ for $v_{\text{proc}}$, not the atmospheric pressure curve.
3. **Semi-crystalline vs amorphous:** Semi-crystalline plastics (PP, POM, PA) show a sharp specific-volume drop at crystallisation — this jump is included in the shrinkage. Amorphous materials (PC, ABS) have a smooth transition and much lower shrinkage.
4. **Gate freeze temperature:** The gate freezes at the thickest cross-section. Use the temperature where the isobar kink/plateau ends (for semi-crystalline) or the glass-transition inflection (for amorphous).

```python
# ═══════════════════════════════════════════════════════════════════════════════
# Part 2 — Problem Inputs  (edit values here)
# ═══════════════════════════════════════════════════════════════════════════════

# ── Read these values from the pvT graph ─────────────────────────────────────
# v_proc:  specific volume at gate-freeze temperature and melt pressure [cm³/g]
# v_solid: specific volume at ambient temperature (23°C) and 1 atm [cm³/g]

# Example: PP from Q11 (T_freeze = 135°C)
v_proc_PP  = 1.30   # cm³/g  — read from pvT at T_freeze on melt-pressure curve
v_solid_PP = 1.22   # cm³/g  — read from pvT at 23°C on atmospheric curve

# Example: PC-ABS from Q11 (T_freeze = 135°C)
v_proc_PCABS  = 0.88   # cm³/g
v_solid_PCABS = 0.86   # cm³/g

# ── Part dimension ────────────────────────────────────────────────────────────
P_d = 50.0   # desired part dimension [mm] for mould cavity calculation

# ── Materials to calculate ───────────────────────────────────────────────────
# List of (label, v_proc, v_solid)
materials = [
    ('PP',     v_proc_PP,    v_solid_PP),
    ('PC-ABS', v_proc_PCABS, v_solid_PCABS),
]
```

```python
# ═══════════════════════════════════════════════════════════════════════════════
# Part 3 — Computation + Validation
# ═══════════════════════════════════════════════════════════════════════════════

print('--- pvT Shrinkage Results ---')
print(f"  {'Material':<10} {'v_proc':<10} {'v_solid':<10} {'S_vol [%]':<12} {'S_lin [%]':<12} {'M_d [mm]':<12}")
print(f"  {'--------':<10} {'------':<10} {'-------':<10} {'---------':<12} {'---------':<12} {'--------':<12}")

for label, v_proc, v_solid in materials:
    S_vol = (v_proc - v_solid) / v_proc * 100.0   # volumetric shrinkage [%]
    S_lin = S_vol / 3.0                            # linear shrinkage [%] (isotropic)
    M_d   = P_d / (1.0 - S_lin / 100.0)           # mould cavity dimension [mm]

    print(f"  {label:<10} {v_proc:<10.3f} {v_solid:<10.3f} {S_vol:<12.2f} {S_lin:<12.2f} {M_d:<12.3f}")

print()
print('--- VALIDATION CHECKS ---')
for label, v_proc, v_solid in materials:
    S_vol = (v_proc - v_solid) / v_proc * 100.0
    S_lin = S_vol / 3.0

    # Sanity checks
    ok_positive  = v_proc > v_solid
    ok_range_vol = 0 < S_vol < 20
    ok_range_lin = 0 < S_lin < 7

    status = 'OK' if (ok_positive and ok_range_vol and ok_range_lin) else 'CHECK INPUTS'
    print(f"  {label:<10}: S_vol={S_vol:.2f}%  S_lin={S_lin:.2f}%  →  {status}")
    if not ok_positive:
        print(f"    ⚠ v_proc must be greater than v_solid (check graph reading)")
    if not ok_range_vol:
        print(f"    ⚠ S_vol={S_vol:.1f}% is outside typical range 0–20%")
```
