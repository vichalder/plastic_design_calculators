```python
inner_diameter = 14.85
shaft = 15
mu = 0.1
length = 30
poisson = 0.4
E_R = 1000*10**6
interference = shaft - inner_diameter
pressure = (interference * E_R)/(inner_diameter * ((inner_diameter**2 + shaft**2) / (inner_diameter**2 - shaft**2) + poisson))

T = mu * pressure * (3.14 * inner_diameter * length) * inner_diameter/2

print(pressure)
print(T)
```
